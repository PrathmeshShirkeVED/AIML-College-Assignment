```python
graph = {
    'A': ['B', 'C'],
    'B': ['D', 'E'],
    'C': ['F'],
    'D': [],
    'E': ['F'],
    'F': []
}

def dfs(graph, start):
  visited = set()
  stack = [start]

  while stack:
    node = stack.pop()

    if node is not visited:
      print(node,end="")
      visited.add(node)

      #Add neighbors in reverse order
      for neighbor in reversed(graph[node]):
        if neighbor not in visited:
          stack.append(neighbor)

dfs(graph, 'A')
```
